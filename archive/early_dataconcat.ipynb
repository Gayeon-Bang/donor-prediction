```python
import pandas as pd
import os

# 경로 및 파일 리스트
base_path = r"C:\Users\bang5\OneDrive\바탕 화면\기부_예측_프로젝트"
file_names = [
    "2023_복지_사회참여_문화와여가_소득과소비_노동_20250807_56404",
    "2021_복지_사회참여_문화와여가_소득과소비_노동_20250807_56404",
    "2019_복지_사회참여_문화와여가_소득과소비_노동_20250807_56404",
    "2017_복지_사회참여_문화와여가_소득과소비_노동_20250807_56404",
    "2015_복지_사회참여_문화와여가_소득과소비_노동_20250807_99967",
    "2013_복지_사회참여_문화와여가_소득과소비_노동_20250807_99967",
    "2011_복지_사회참여_문화와여가_소득과소비_노동_20250807_99967"
]

# 모든 데이터 로딩
data_by_year = {}
columns_2023 = []

for file in file_names:
    year = int(file[:4])
    path = os.path.join(base_path, f"{file}.csv")

    if not os.path.exists(path):
        print(f"[경고] 파일 없음: {path}")
        continue

    try:
        try:
            df = pd.read_csv(path, encoding='utf-8')
        except UnicodeDecodeError:
            df = pd.read_csv(path, encoding='cp949')

        df['년도'] = year
        data_by_year[year] = df

        if year == 2023:
            columns_2023 = [col for col in df.columns if col != '년도']
            print(f"[기준] 2023년 컬럼 확보 완료 ({len(columns_2023)}개)")
    except Exception as e:
        print(f"[오류] {file}.csv 로딩 실패: {e}")

# 각 컬럼별 세로 병합 후 전체 가로 결합
column_wise_blocks = []
year_column = []

for col in columns_2023:
    temp_list = []
    for year, df in data_by_year.items():
        if col in df.columns:
            temp = df[[col, '년도']].copy()
            temp_list.append(temp)
    if temp_list:
        block = pd.concat(temp_list, axis=0, ignore_index=True)
        column_wise_blocks.append(block[[col]])
        year_column.append(block['년도'])  # 동일한 순서일 것 가정
        print(f"[병합] '{col}' 컬럼 세로 결합 완료 (rows: {block.shape[0]})")
    else:
        print(f"[스킵] '{col}' → 해당 컬럼 가진 연도 없음")

# 최종 가로 결합
final_df = pd.concat(column_wise_blocks, axis=1)

# 마지막에 '년도' 컬럼 1개만 추가
if year_column:
    final_df.insert(0, '년도', year_column[0].reset_index(drop=True))

# 결과 확인
print("\n✅ 최종 결합 shape:", final_df.shape)
print(final_df.head())

# 저장 옵션
# final_df.to_csv("기부_2023기준_순차결합.csv", index=False, encoding='utf-8-sig')
```

```python
import pandas as pd
import os

# 연도순 정렬
years = [2023, 2021, 2019, 2017, 2015, 2013, 2011]
file_map = {
    2023: "2023_복지_사회참여_문화와여가_소득과소비_노동_20250807_56404",
    2021: "2021_복지_사회참여_문화와여가_소득과소비_노동_20250807_56404",
    2019: "2019_복지_사회참여_문화와여가_소득과소비_노동_20250807_56404",
    2017: "2017_복지_사회참여_문화와여가_소득과소비_노동_20250807_56404",
    2015: "2015_복지_사회참여_문화와여가_소득과소비_노동_20250807_99967",
    2013: "2013_복지_사회참여_문화와여가_소득과소비_노동_20250807_99967",
    2011: "2011_복지_사회참여_문화와여가_소득과소비_노동_20250807_99967",
}

base_path = r"C:\Users\bang5\OneDrive\바탕 화면\기부_예측_프로젝트"
data_by_year = {}

# 1. 모든 연도 데이터 미리 불러오기
for year in years:
    path = os.path.join(base_path, file_map[year] + ".csv")
    if not os.path.exists(path):
        print(f"[경고] 파일 없음: {path}")
        continue
    try:
        try:
            df = pd.read_csv(path, encoding='utf-8')
        except UnicodeDecodeError:
            df = pd.read_csv(path, encoding='cp949')
        df['년도'] = year
        data_by_year[year] = df
        print(f"[성공] {year}년 데이터 로딩 완료: shape={df.shape}")
    except Exception as e:
        print(f"[오류] {year}년 데이터 로딩 실패: {e}")

# 2. 순차 결합 로직
final_df = data_by_year[2023].copy()

for year in years[1:]:
    print(f"\n {year}년 데이터 결합 시작")

    if year not in data_by_year:
        print(f"[건너뜀] {year}년 데이터 없음")
        continue

    current_df = data_by_year[year]

    base_cols = set(final_df.columns) - {'년도'}
    new_cols = set(current_df.columns) - {'년도'}

    common_cols = list(base_cols & new_cols)
    non_common_cols = list(new_cols - base_cols)

    # 1) 공통 컬럼은 세로 결합
    if common_cols:
        df_common = final_df[common_cols + ['년도']].copy()
        current_common = current_df[common_cols + ['년도']].copy()
        final_df = pd.concat([df_common, current_common], axis=0, ignore_index=True)
        print(f"공통 컬럼 세로 결합 완료: {len(common_cols)}개 컬럼")
    else:
        print("공통 컬럼 없음 → 세로 결합 생략")

    # 2) 비공통 컬럼은 가로 결합 (NaN 채움)
    if non_common_cols:
        extra_part = current_df[non_common_cols].copy()
        extra_part = extra_part.reset_index(drop=True)

        # final_df와 행 수 맞추기
        row_diff = len(final_df) - len(extra_part)
        if row_diff > 0:
            extra_part = pd.concat([extra_part, pd.DataFrame(index=range(row_diff))], ignore_index=True)
        elif row_diff < 0:
            final_df = pd.concat([final_df, pd.DataFrame(index=range(-row_diff))], ignore_index=True)

        final_df = pd.concat([final_df.reset_index(drop=True), extra_part.reset_index(drop=True)], axis=1)
        print(f"비공통 컬럼 가로 결합 완료: {len(non_common_cols)}개 컬럼")

    else:
        print("가로로 붙일 컬럼 없음")

print("\n모든 연도 결합 완료 → 최종 shape:", final_df.shape)
# 저장 선택 가능
# final_df.to_csv("기부_순차결합완료.csv", index=False, encoding='utf-8-sig')
```

```python
import pandas as pd
import os

# 연도별 파일 매핑
years = [2023, 2021, 2019, 2017, 2015, 2013, 2011]
file_map = {
    2023: "2023_복지_사회참여_문화와여가_소득과소비_노동_20250807_56404",
    2021: "2021_복지_사회참여_문화와여가_소득과소비_노동_20250807_56404",
    2019: "2019_복지_사회참여_문화와여가_소득과소비_노동_20250807_56404",
    2017: "2017_복지_사회참여_문화와여가_소득과소비_노동_20250807_56404",
    2015: "2015_복지_사회참여_문화와여가_소득과소비_노동_20250807_99967",
    2013: "2013_복지_사회참여_문화와여가_소득과소비_노동_20250807_99967",
    2011: "2011_복지_사회참여_문화와여가_소득과소비_노동_20250807_99967"
}

base_path = r"C:\Users\bang5\OneDrive\바탕 화면\기부_예측_프로젝트"
data_by_year = {}

# 1. 모든 연도 데이터 미리 로딩
for year in years:
    path = os.path.join(base_path, file_map[year] + ".csv")
    if not os.path.exists(path):
        print(f"[경고] {year}년 파일 없음: {path}")
        continue
    try:
        try:
            df = pd.read_csv(path, encoding='utf-8')
        except UnicodeDecodeError:
            df = pd.read_csv(path, encoding='cp949')
        df['년도'] = year
        data_by_year[year] = df
        print(f"[성공] {year}년 데이터 로딩 완료: shape={df.shape}")
    except Exception as e:
        print(f"[오류] {year}년 데이터 로딩 실패: {e}")

# 2. 순차 결합 (칼럼 절대 삭제 X)
final_df = data_by_year[2023].copy()

for year in years[1:]:
    print(f"\n {year}년 데이터 결합 시작")

    if year not in data_by_year:
        print(f"[건너뜀] {year}년 데이터 없음")
        continue

    current_df = data_by_year[year]

    final_cols = set(final_df.columns)
    current_cols = set(current_df.columns)

    # 공통 컬럼
    common_cols = list(final_cols & current_cols)
    # 추가되는 컬럼 (final에는 없고, current에만 있음)
    new_cols = list(current_cols - final_cols)
    # final에만 있고 current에 없는 컬럼
    missing_cols = list(final_cols - current_cols)

    # 1. current_df에 final_df에 없는 컬럼 추가 → NaN 채움
    for col in missing_cols:
        current_df[col] = pd.NA

    # 2. final_df에 current_df에 없는 컬럼 추가 → NaN 채움
    for col in new_cols:
        final_df[col] = pd.NA

    # 3. 컬럼 순서를 맞추기 위해 정렬
    current_df = current_df[final_df.columns]

    # 4. 세로 결합 (절대 칼럼 삭제 없이)
    final_df = pd.concat([final_df, current_df], axis=0, ignore_index=True)
    print(f" {year}년 데이터 병합 완료 → 누적 shape: {final_df.shape}")

# 최종 확인
print("\n 전체 데이터 결합 완료")
print("최종 shape:", final_df.shape)
print("전체 컬럼 수:", len(final_df.columns))
print(final_df.head())

# 저장 예시
# final_df.to_csv("기부_통합_전칼럼유지.csv", index=False, encoding='utf-8-sig')
```

```python
import pandas as pd
import os

# 연도별 파일 매핑
years = [2023, 2021, 2019, 2017, 2015]
file_map = {
    2023: "2023_복지_사회참여_문화와여가_소득과소비_노동_20250807_56404",
    2021: "2021_복지_사회참여_문화와여가_소득과소비_노동_20250807_56404",
    2019: "2019_복지_사회참여_문화와여가_소득과소비_노동_20250807_56404",
    2017: "2017_복지_사회참여_문화와여가_소득과소비_노동_20250807_56404",
    2015: "2015_복지_사회참여_문화와여가_소득과소비_노동_20250807_99967"
}

base_path = r"C:\Users\bang5\OneDrive\바탕 화면\기부_예측_프로젝트"
data_by_year = {}

# 1. 모든 연도 데이터 미리 로딩
for year in years:
    path = os.path.join(base_path, file_map[year] + ".csv")
    if not os.path.exists(path):
        print(f"[경고] {year}년 파일 없음: {path}")
        continue
    try:
        try:
            df = pd.read_csv(path, encoding='utf-8')
        except UnicodeDecodeError:
            df = pd.read_csv(path, encoding='cp949')
        df['년도'] = year
        data_by_year[year] = df
        print(f"[성공] {year}년 데이터 로딩 완료: shape={df.shape}")
    except Exception as e:
        print(f"[오류] {year}년 데이터 로딩 실패: {e}")

# 2. 순차 결합 (칼럼 절대 삭제 X)
final_df = data_by_year[2023].copy()

for year in years[1:]:
    print(f"\n {year}년 데이터 결합 시작")

    if year not in data_by_year:
        print(f"[건너뜀] {year}년 데이터 없음")
        continue

    current_df = data_by_year[year]

    final_cols = set(final_df.columns)
    current_cols = set(current_df.columns)

    # 공통 컬럼
    common_cols = list(final_cols & current_cols)
    # 추가되는 컬럼 (final에는 없고, current에만 있음)
    new_cols = list(current_cols - final_cols)
    # final에만 있고 current에 없는 컬럼
    missing_cols = list(final_cols - current_cols)

    # 1. current_df에 final_df에 없는 컬럼 추가 → NaN 채움
    for col in missing_cols:
        current_df[col] = pd.NA

    # 2. final_df에 current_df에 없는 컬럼 추가 → NaN 채움
    for col in new_cols:
        final_df[col] = pd.NA

    # 3. 컬럼 순서를 맞추기 위해 정렬
    current_df = current_df[final_df.columns]

    # 4. 세로 결합 (절대 칼럼 삭제 없이)
    final_df = pd.concat([final_df, current_df], axis=0, ignore_index=True)
    print(f" {year}년 데이터 병합 완료 → 누적 shape: {final_df.shape}")

# 최종 확인
print("\n 전체 데이터 결합 완료")
print("최종 shape:", final_df.shape)
print("전체 컬럼 수:", len(final_df.columns))
print(final_df.head())
```
