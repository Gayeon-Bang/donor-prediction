```python
import pandas as pd
import numpy as np
from sklearn.preprocessing import LabelEncoder
from sklearn.impute import SimpleImputer

# 1. 파일 불러오기
df = pd.read_csv("df_merged_preprocessed.csv", low_memory=False)

target_col = "기부여부"

# 2. 결측치 50% 이상 컬럼 삭제
df = df.drop(columns=df.columns[df.isnull().mean() >= 0.5])

# 3. 나머지 결측치는 최빈값으로 대체
df = pd.DataFrame(SimpleImputer(strategy="most_frequent").fit_transform(df), columns=df.columns)

# 4. 타겟 정수형 변환
df[target_col] = df[target_col].astype(int)

# 5. 입력(X), 타겟(y) 분리
X = df.drop(columns=[target_col])
y = df[target_col]

# 6. 모든 범주형 변수 라벨 인코딩 (원핫 인코딩 안 함 → 차원 폭발 방지)
for col in X.select_dtypes(include=["object", "category"]).columns:
    le = LabelEncoder()
    X[col] = le.fit_transform(X[col].astype(str))

# 7. 저장
X.to_csv("X_preprocessed_simple.csv", index=False)
y.to_csv("y_preprocessed_simple.csv", index=False)

print("전처리 완료: X_preprocessed_simple.csv, y_preprocessed_simple.csv 저장됨")
```
