```python

# 1. 데이터 전처리 + 저장 
# ===============================
import pandas as pd
import numpy as np
from sklearn.preprocessing import MinMaxScaler
from sklearn.impute import SimpleImputer
from sklearn.feature_selection import VarianceThreshold

# 1-1. 통합 전처리된 파일 불러오기
df = pd.read_csv("df_merged_preprocessed.csv", low_memory=False)

target_col = '기부여부'

# 1-2. 결측치 50% 이상 컬럼 삭제
df = df.drop(columns=df.columns[df.isnull().mean() >= 0.5])

# 1-3. 나머지 결측치는 최빈값으로 대체
df = pd.DataFrame(SimpleImputer(strategy='most_frequent').fit_transform(df), columns=df.columns)

# 1-4. 타겟 변수 정수형 변환
df[target_col] = df[target_col].astype(int)

# 1-5. 입력(X), 타겟(y) 분리
X = df.drop(columns=[target_col])
y = df[target_col]

# 1-6. 모든 범주형(object) 변수 원-핫 인코딩
X = pd.get_dummies(X, drop_first=True)

# 1-7. 분산이 거의 없는 변수 제거
X = pd.DataFrame(VarianceThreshold(threshold=0.01).fit_transform(X))

# 1-8. 다중공선성 제거 (상관계수 > 0.9인 컬럼 제거)
corr_matrix = X.corr().abs()
upper = corr_matrix.where(np.triu(np.ones(corr_matrix.shape), k=1).astype(bool))
drop_cols = [col for col in upper.columns if any(upper[col] > 0.9)]
X.drop(X.columns[drop_cols], axis=1, inplace=True)

# 1-9. 전처리 결과 저장
X.to_csv("X_preprocessed.csv", index=False)
y.to_csv("y_preprocessed.csv", index=False)

print("전처리 완료: X_preprocessed.csv, y_preprocessed.csv 저장됨")
```

```python
##젘디널리티만. 

import pandas as pd
import numpy as np
from sklearn.preprocessing import MinMaxScaler
from sklearn.impute import SimpleImputer
from sklearn.feature_selection import VarianceThreshold

# 1-1. 통합 전처리된 파일 불러오기
df = pd.read_csv("df_merged_preprocessed.csv", low_memory=False)

target_col = '기부여부'

# 1-2. 결측치 50% 이상 컬럼 삭제
df = df.drop(columns=df.columns[df.isnull().mean() >= 0.5])

# 1-3. 나머지 결측치는 최빈값으로 대체
df = pd.DataFrame(SimpleImputer(strategy='most_frequent').fit_transform(df), columns=df.columns)

# 1-4. 타겟 변수 정수형 변환
df[target_col] = df[target_col].astype(int)

# 1-5. 입력(X), 타겟(y) 분리
X = df.drop(columns=[target_col])
y = df[target_col]

# 1-6. 범주형 처리 - 저카디널리티만 더미화
low_card_cols = [c for c in X.select_dtypes(include=["object", "category"]).columns if X[c].nunique() <= 20]
high_card_cols = [c for c in X.select_dtypes(include=["object", "category"]).columns if X[c].nunique() > 20]

# 저카디널리티 → 원핫 인코딩
X = pd.get_dummies(X, columns=low_card_cols, drop_first=True)

# 고카디널리티 → 라벨 인코딩
from sklearn.preprocessing import LabelEncoder
for col in high_card_cols:
    le = LabelEncoder()
    X[col] = le.fit_transform(X[col].astype(str))


# 1-7. 분산이 거의 없는 변수 제거
X = pd.DataFrame(VarianceThreshold(threshold=0.01).fit_transform(X))

# 1-8. 다중공선성 제거 (상관계수 > 0.9인 컬럼 제거)
corr_matrix = X.corr().abs()
upper = corr_matrix.where(np.triu(np.ones(corr_matrix.shape), k=1).astype(bool))
drop_cols = [col for col in upper.columns if any(upper[col] > 0.9)]
X.drop(X.columns[drop_cols], axis=1, inplace=True)

# 1-9. 전처리 결과 저장
X.to_csv("X_preprocessed.csv", index=False)
y.to_csv("y_preprocessed.csv", index=False)

print("전처리 완료: X_preprocessed.csv, y_preprocessed.csv 저장됨")
```

```python
import pandas as pd
import numpy as np
from sklearn.preprocessing import LabelEncoder
from sklearn.impute import SimpleImputer

# 1. 파일 불러오기
df = pd.read_csv("df_merged_preprocessed.csv", low_memory=False)

target_col = "기부여부"

# 2. 결측치 50% 이상 컬럼 삭제
df = df.drop(columns=df.columns[df.isnull().mean() >= 0.5])

# 3. 나머지 결측치는 최빈값으로 대체
df = pd.DataFrame(SimpleImputer(strategy="most_frequent").fit_transform(df), columns=df.columns)

# 4. 타겟 정수형 변환
df[target_col] = df[target_col].astype(int)

# 5. 입력(X), 타겟(y) 분리
X = df.drop(columns=[target_col])
y = df[target_col]

# 6. 모든 범주형 변수 라벨 인코딩 (원핫 인코딩 안 함 → 차원 폭발 방지)
for col in X.select_dtypes(include=["object", "category"]).columns:
    le = LabelEncoder()
    X[col] = le.fit_transform(X[col].astype(str))

# 7. 저장
X.to_csv("X_preprocessed_simple.csv", index=False)
y.to_csv("y_preprocessed_simple.csv", index=False)

print("전처리 완료: X_preprocessed_simple.csv, y_preprocessed_simple.csv 저장됨")
```
