```python
# -*- coding: utf-8 -*-
import pandas as pd
import numpy as np

# ---------- 0) Load ----------
path = "df_merged_preprocessed.csv"
df = pd.read_csv(path, low_memory=False)

print("File loaded:", path)
print(f"Shape: {df.shape[0]:,} rows × {df.shape[1]:,} columns\n")

# ---------- 1) Column-level summary ----------
summary_list = []

for col in df.columns:
    dtype = str(df[col].dtype)
    n_missing = df[col].isna().sum()
    missing_pct = round((n_missing / len(df)) * 100, 3)
    n_unique = df[col].nunique(dropna=True)

    # 기본값
    stats = {
        "column": col,
        "dtype": dtype,
        "n_missing": n_missing,
        "missing_pct": missing_pct,
        "n_unique": n_unique,
        "min": None,
        "p05": None,
        "median": None,
        "p95": None,
        "max": None,
    }

    # 수치형일 경우만 통계 계산
    if np.issubdtype(df[col].dtype, np.number):
        stats["min"] = df[col].min(skipna=True)
        stats["p05"] = df[col].quantile(0.05)
        stats["median"] = df[col].median(skipna=True)
        stats["p95"] = df[col].quantile(0.95)
        stats["max"] = df[col].max(skipna=True)

    summary_list.append(stats)

summary_df = pd.DataFrame(summary_list)

# 보기 좋게 저장
summary_path = "merged_column_summary.csv"
summary_df.to_csv(summary_path, index=False, encoding="utf-8-sig")
```
